# Задачи по вероятности

Решения заданий из `probability_tasks.pdf`. Номера задач соответствуют исходному документу; здесь они расположены по возрастанию.

## 1. Монеты

Честную монету бросили два раза независимо. Какова вероятность, что оба раза выпал орёл?

Вероятность орла при одном броске равна $1/2$. Броски независимы, поэтому вероятности перемножаю:

$$P(\text{два орла})=\frac12\cdot\frac12=\frac14.$$

**Ответ:** $\boxed{1/4=0.25}$.

## 2. Монета и кубик

Независимо бросают честную монету и обычный игральный кубик. Какова вероятность, что выпадут орёл и шестёрка?

Вероятность орла равна $1/2$, вероятность шестёрки — $1/6$. Оба события должны произойти одновременно и не зависят друг от друга:

$$P(\text{орёл и шестёрка})=\frac12\cdot\frac16=\frac1{12}.$$

**Ответ:** $\boxed{1/12\approx0.08333}$.

## 3. Два стрелка

Первый стрелок попадает с вероятностью 0,8, второй — 0,5. Они стреляют независимо по одному разу. Найдите вероятность, что: а) оба попадут; б) хотя бы один попадёт.

а) Для двух попаданий перемножаю вероятности независимых событий:

$$P(\text{оба попадут})=0.8\cdot0.5=0.4.$$

б) «Хотя бы один попадёт» означает, что попадёт один или оба. Удобнее исключить случай, когда оба промахнулись. Вероятности промаха равны $0.2$ и $0.5$:

$$P(\text{хотя бы один})=1-(1-0.8)(1-0.5)=1-0.1=0.9.$$

**Ответ:** а) $\boxed{0.4}$; б) $\boxed{0.9}$.

## 4. Две коробки

В первой коробке 3 красных и 1 синий шар, во второй — 1 красный и 3 синих. Коробку выбирают равновероятно, затем случайно достают шар. Он оказался красным. Какова вероятность, что выбрали первую коробку?

Обозначаю выбор первой коробки через $B_1$, второй — через $B_2$, красный шар — через $R$. Сначала нахожу вероятность достать красный шар любым способом:

$$P(R)=P(B_1)P(R\mid B_1)+P(B_2)P(R\mid B_2)
=\frac12\cdot\frac34+\frac12\cdot\frac14=\frac12.$$

Теперь учитываю, что шар уже оказался красным. По формуле Байеса:

$$P(B_1\mid R)=\frac{P(B_1)P(R\mid B_1)}{P(R)}
=\frac{\frac12\cdot\frac34}{\frac12}=\frac34.$$

Красный шар в три раза вероятнее получить из первой коробки, чем из второй, при одинаковой вероятности выбора коробок.

**Ответ:** $\boxed{3/4=0.75}$.

## 5. Необычная монета

Есть две монеты: одна честная, у другой орёл с обеих сторон. Случайно выбирают одну из них с равными вероятностями и бросают. Выпал орёл. Какова вероятность, что выбрали необычную монету?

Обозначаю необычную монету через $U$, выпадение орла — через $H$. Для необычной монеты $P(H\mid U)=1$, для честной — $1/2$. Общая вероятность орла:

$$P(H)=\frac12\cdot1+\frac12\cdot\frac12=\frac34.$$

После выпадения орла вероятность выбора необычной монеты равна:

$$P(U\mid H)=\frac{P(U)P(H\mid U)}{P(H)}
=\frac{\frac12\cdot1}{\frac34}=\frac23.$$

Наблюдение орла делает необычную монету более вероятной: она всегда даёт орла, а честная — только в половине случаев.

**Ответ:** $\boxed{2/3\approx0.66667}$.

## 6. Спам-фильтр

Половина писем — спам. Фильтр помечает 80% спама и ошибочно помечает 20% обычных писем. Письмо помечено фильтром. Какова вероятность, что это спам?

Для наглядности рассматриваю 100 писем с указанными долями. Из 50 спам-писем фильтр помечает $50\cdot0.8=40$, из 50 обычных — $50\cdot0.2=10$.

Среди 50 помеченных писем 40 являются спамом:

$$P(\text{спам}\mid\text{помечено})
=\frac{0.5\cdot0.8}{0.5\cdot0.8+0.5\cdot0.2}
=\frac{40}{40+10}=0.8.$$

**Ответ:** $\boxed{0.8=80\%}$.

## 7. Ровно одно попадание

Три стрелка независимо делают по одному выстрелу. Вероятности попадания — 0,5, 0,6 и 0,8. Какова вероятность, что попадёт ровно один стрелок?

Подходят три несовместимых случая: попал только первый, только второй или только третий. Для каждого перемножаю вероятность нужного попадания и двух промахов, затем складываю:

$$\begin{aligned}
P(\text{только первый})&=0.5\cdot0.4\cdot0.2=0.04,\\
P(\text{только второй})&=0.5\cdot0.6\cdot0.2=0.06,\\
P(\text{только третий})&=0.5\cdot0.4\cdot0.8=0.16.
\end{aligned}$$

$$P(\text{ровно одно попадание})=0.04+0.06+0.16=0.26.$$

**Ответ:** $\boxed{0.26}$.

## 8. Работа системы

Устройство работает, если исправен хотя бы один из двух блоков. Каждый блок состоит из двух деталей и исправен, только если обе его детали исправны. Каждая из четырёх деталей исправна с вероятностью 0,9, независимо от остальных. Найдите вероятность работы устройства.

Один блок работает, только когда исправны обе его детали:

$$P(\text{блок работает})=0.9\cdot0.9=0.81.$$

Вероятность отказа блока равна $1-0.81=0.19$. Блоки состоят из разных деталей, а состояния всех деталей независимы, поэтому отказы блоков тоже независимы.

Устройство откажет только при отказе обоих блоков. Вычитаю вероятность этого случая из единицы:

$$P(\text{устройство работает})=1-0.19^2=1-0.0361=0.9639.$$

**Ответ:** $\boxed{0.9639}$.

## 9. Известно, что кто-то попал

Два стрелка независимо делают по одному выстрелу, попадая с вероятностями 0,8 и 0,6. Известно, что хотя бы один попал. Какова вероятность, что попали оба?

Пусть $A$ — оба попали, $B$ — хотя бы один попал. Событие $A$ входит в $B$, поэтому $A\cap B=A$.

$$P(A)=0.8\cdot0.6=0.48,$$

$$P(B)=1-(1-0.8)(1-0.6)=1-0.2\cdot0.4=0.92.$$

Уже известно, что произошло $B$, поэтому рассматриваю только такие исходы:

$$P(A\mid B)=\frac{P(A\cap B)}{P(B)}
=\frac{0.48}{0.92}=\frac{12}{23}\approx0.52174.$$

**Ответ:** $\boxed{12/23\approx0.52174}$.

## 10. Дискретное распределение

Случайная величина $X$ принимает значения $-2$, $0$, $3$, $5$ с вероятностями $0.2$, $0.3$, $0.4$, $0.1$ соответственно. Найдите $E[X]$ и $E[2X+1]$.

Математическое ожидание дискретной величины — сумма её значений с весами, равными вероятностям:

$$E[X]=(-2)\cdot0.2+0\cdot0.3+3\cdot0.4+5\cdot0.1
=-0.4+1.2+0.5=1.3.$$

Использую линейность математического ожидания:

$$E[2X+1]=2E[X]+1=2\cdot1.3+1=3.6.$$

**Ответ:** $\boxed{E[X]=1.3}$, $\boxed{E[2X+1]=3.6}$.

## 11. Равномерная плотность

Случайная величина $X$ имеет плотность $f(x)=1/4$ при $2\le x\le6$ и $f(x)=0$ вне этого отрезка. Найдите $E[X]$.

Плотность постоянна на отрезке $[2,6]$, поэтому распределение равномерное. Его среднее находится в середине отрезка: $(2+6)/2=4$.

Это же получается из определения через интеграл:

$$E[X]=\int_2^6 x\cdot\frac14\,dx
=\left.\frac{x^2}{8}\right|_2^6
=\frac{36-4}{8}=4.$$

**Ответ:** $\boxed{E[X]=4}$.

## 12. Плотность с неизвестной константой

Случайная величина $X$ имеет плотность $f(x)=cx$ при $0\le x\le2$ и $f(x)=0$ вне этого отрезка. Найдите константу $c$, затем $E[X]$.

Интеграл плотности по всей области должен равняться единице:

$$\int_0^2 cx\,dx
=\left.\frac{cx^2}{2}\right|_0^2=2c=1.$$

Отсюда $c=1/2$. При таком $c$ плотность также неотрицательна на заданном отрезке.

Для ожидания умножаю плотность на $x$ и интегрирую:

$$E[X]=\int_0^2 x\cdot\frac{x}{2}\,dx
=\left.\frac{x^3}{6}\right|_0^2=\frac86=\frac43.$$

**Ответ:** $\boxed{c=1/2}$, $\boxed{E[X]=4/3}$.

## 13. Урон роги

Рога бросает честный кубик d20. При результате 1–15 он промахивается и наносит 0 урона; при 16–19 попадает и наносит 4d6 урона; при 20 критует и удваивает количество кубиков: урон равен 8d6. Обозначение nd6 означает сумму результатов $n$ честных шестигранных кубиков. Все броски независимы, дополнительных модификаторов урона нет. Найдите математическое ожидание урона за одну атаку.

Среднее значение одного шестигранного кубика:

$$E[d6]=\frac{1+2+3+4+5+6}{6}=3.5.$$

По линейности ожидания средний урон при обычном попадании равен $4\cdot3.5=14$, при критическом — $8\cdot3.5=28$.

На d20 есть 15 исходов промаха, 4 исхода обычного попадания и 1 исход критического. Взвешиваю средний урон каждого случая по его вероятности:

$$E[D]=\frac{15}{20}\cdot0+\frac4{20}\cdot14+\frac1{20}\cdot28
=2.8+1.4=4.2.$$

Это средний урон за одну атаку с учётом промахов, а не только при условии попадания.

**Ответ:** $\boxed{E[D]=4.2}$.

## 14. Математическое ожидание через интеграл

Случайная величина $X$ имеет плотность $f(x)=3x^2$ при $0\le x\le1$ и $f(x)=0$ вне этого отрезка. Найдите математическое ожидание $E[X]$, вычислив интеграл.

По определению математического ожидания непрерывной величины интегрирую $x f(x)$. Вне отрезка $[0,1]$ плотность равна нулю:

$$E[X]=\int_0^1 x\cdot3x^2\,dx
=\int_0^1 3x^3\,dx
=\left.\frac{3x^4}{4}\right|_0^1
=\frac34.$$

**Ответ:** $\boxed{E[X]=3/4=0.75}$.
